# Outlier Detection Using IQR Method

```python
Q1 = df[numerical_columns].quantile(0.25)
Q3 = df[numerical_columns].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - (1.5 * IQR)
upper_limit = Q3 + (1.5 * IQR)

outlier_count = (
    (df[numerical_columns] < lower_limit) |
    (df[numerical_columns] > upper_limit)
).sum()

display(outlier_count.to_frame(name="Outlier Count"))
```

# Outlier Visualization Using Boxplots

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(16, 8))

df[numerical_columns].boxplot(rot=45)

plt.title("Outlier Detection Across Numerical Features")
plt.ylabel("Values")
plt.tight_layout()
plt.show()
```

# Individual Outlier Inspection

```python
outlier_details = {}

for column in numerical_columns:
    lower = lower_limit[column]
    upper = upper_limit[column]

    outliers = df.loc[
        (df[column] < lower) | (df[column] > upper),
        column
    ]

    outlier_details[column] = outliers.values

    if len(outliers) > 0:
        print(f"\n{column}")
        print(outliers.to_list())
```

# Date and Time Consistency Check

```python
# Convert charging start time into datetime format

df["Charging Start Time"] = pd.to_datetime(
    df["Charging Start Time"]
)

# Extract actual day name from date

actual_day = df["Charging Start Time"].dt.day_name()

# Compare actual day with existing day column

day_mismatch = df[
    actual_day != df["Day of Week"]
]

print("Day of Week Mismatches:", len(day_mismatch))

display(
    day_mismatch[
        ["Charging Start Time", "Day of Week"]
    ].head(10)
)
```

# Time of Day Validation

```python
# Extract hour from charging start time

hour = df["Charging Start Time"].dt.hour

# Assign time period based on actual hour

actual_time_period = pd.cut(
    hour,
    bins=[-1, 5, 11, 17, 23],
    labels=["Night", "Morning", "Afternoon", "Evening"]
)

# Compare with existing Time of Day column

time_mismatch = df[
    actual_time_period.astype(str) != df["Time of Day"]
]

print("Time of Day Mismatches:", len(time_mismatch))

display(
    time_mismatch[
        ["Charging Start Time", "Time of Day"]
    ].head(10)
)
```

# Date and Time Mismatch Summary

```python
print("Total Records:", len(df))

print("Day of Week Mismatches:", len(day_mismatch))

print("Time of Day Mismatches:", len(time_mismatch))

print("Day Mismatch Percentage:",
      round(len(day_mismatch) / len(df) * 100, 2), "%")

print("Time Mismatch Percentage:",
      round(len(time_mismatch) / len(df) * 100, 2), "%")
```

# Missing Values Imputation

```python
for column in missing_columns:
    median_value = df_clean[column].median()
    df_clean[column] = df_clean[column].fillna(median_value)

print("Missing Values After Imputation:")
display(df_clean[missing_columns].isnull().sum())
```

# State of Charge (SOC) Validation

```python
soc_columns = [
    "State of Charge (Start %)",
    "State of Charge (End %)"
]

for column in soc_columns:
    invalid_values = df_clean[
        (df_clean[column] < 0) | (df_clean[column] > 100)
    ]

    print(f"\n{column}")
    print("Invalid Records:", len(invalid_values))
    display(invalid_values[[column]].head(10))
```

# Invalid SOC Records Inspection

```python
invalid_soc_records = df_clean[
    (df_clean["State of Charge (Start %)"] < 0) |
    (df_clean["State of Charge (Start %)"] > 100) |
    (df_clean["State of Charge (End %)"] < 0) |
    (df_clean["State of Charge (End %)"] > 100)
]

print("Total Records with Invalid SOC:", len(invalid_soc_records))

display(invalid_soc_records[
    [
        "User ID",
        "Vehicle Model",
        "Battery Capacity (kWh)",
        "Energy Consumed (kWh)",
        "State of Charge (Start %)",
        "State of Charge (End %)"
    ]
])
```

# Handling Invalid State of Charge Values

```python
for column in soc_columns:
    invalid_mask = (
        (df_clean[column] < 0) |
        (df_clean[column] > 100)
    )

    df_clean.loc[invalid_mask, column] = np.nan

print("Invalid SOC Values After Treatment:")

for column in soc_columns:
    invalid_count = (
        (df_clean[column] < 0) |
        (df_clean[column] > 100)
    ).sum()

    print(f"{column}: {invalid_count}")
```

# Charging Duration Consistency Check

```python
df_clean["Calculated Duration (hours)"] = (
    df_clean["Charging End Time"] -
    df_clean["Charging Start Time"]
).dt.total_seconds() / 3600

df_clean["Duration Difference"] = (
    df_clean["Charging Duration (hours)"] -
    df_clean["Calculated Duration (hours)"]
)

display(
    df_clean[
        [
            "Charging Start Time",
            "Charging End Time",
            "Charging Duration (hours)",
            "Calculated Duration (hours)",
            "Duration Difference"
        ]
    ].head(15)
)
```

# Overall Charging Duration Consistency Analysis

```python
total_records = len(df_clean)

matching_records = (
    df_clean["Duration Difference"].abs() <= 0.05
).sum()

inconsistent_records = (
    df_clean["Duration Difference"].abs() > 0.05
).sum()

print("Total Records:", total_records)
print("Matching Records:", matching_records)
print("Inconsistent Records:", inconsistent_records)

print(
    "Matching Percentage:",
    round((matching_records / total_records) * 100, 2), "%"
)

print(
    "Inconsistent Percentage:",
    round((inconsistent_records / total_records) * 100, 2), "%"
)
```

# Cost per kWh Outlier Detection

```python
Q1_cost = df_clean["Cost per kWh"].quantile(0.25)
Q3_cost = df_clean["Cost per kWh"].quantile(0.75)

IQR_cost = Q3_cost - Q1_cost

lower_limit_cost = Q1_cost - (1.5 * IQR_cost)
upper_limit_cost = Q3_cost + (1.5 * IQR_cost)

cost_outliers = df_clean[
    (df_clean["Cost per kWh"] < lower_limit_cost) |
    (df_clean["Cost per kWh"] > upper_limit_cost)
]

print("Q1:", Q1_cost)
print("Q3:", Q3_cost)
print("IQR:", IQR_cost)
print("Lower Limit:", lower_limit_cost)
print("Upper Limit:", upper_limit_cost)
print("Total Cost per kWh Outliers:", len(cost_outliers))
```

# Cost per kWh Outlier Investigation

```python
cost_outlier_details = df_clean.loc[
    cost_outliers.index,
    [
        "User ID",
        "Vehicle Model",
        "Charging Station ID",
        "Charging Station Location",
        "Energy Consumed (kWh)",
        "Charging Cost (USD)",
        "Cost per kWh"
    ]
]

display(
    cost_outlier_details.sort_values(
        by="Cost per kWh",
        ascending=False
    ).head(10)
)
```

# Cost per kWh Outliers by Vehicle Model

```python
vehicle_outlier_counts = (
    cost_outlier_details["Vehicle Model"]
    .value_counts()
)

display(vehicle_outlier_counts)
```

# Cost per kWh Outliers by Charging Station

```python
station_outlier_counts = (
    cost_outlier_details["Charging Station ID"]
    .value_counts()
)

display(station_outlier_counts.head(10))
```

# Cost Outlier Rate by Charging Station

```python
station_total_counts = df_clean["Charging Station ID"].value_counts()

station_outlier_rate = (
    cost_outlier_details["Charging Station ID"].value_counts()
    .div(station_total_counts)
    .mul(100)
)

station_outlier_summary = pd.DataFrame({
    "Total Records": station_total_counts,
    "Outlier Records": cost_outlier_details["Charging Station ID"].value_counts(),
    "Outlier Rate (%)": station_outlier_rate
})

station_outlier_summary = (
    station_outlier_summary
    .fillna(0)
    .sort_values("Outlier Rate (%)", ascending=False)
)

display(station_outlier_summary.head(10))
```

# Compare Stations with Sufficient Records

```python
reliable_station_summary = station_outlier_summary[
    station_outlier_summary["Total Records"] >= 5
]

reliable_station_summary = reliable_station_summary.sort_values(
    "Outlier Rate (%)",
    ascending=False
)

display(reliable_station_summary.head(10))
```
